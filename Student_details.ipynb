```python
# To removes dublicates and fix invalid values and handle missing values.

#To remove duplicates
df = df.drop_duplicates()
print(df.shape)
```

```python
# To fix invalid values

df.loc[df['study_hours'] < 0, 'study_hours'] = np.nan
print(df['study_hours'].describe())
```

```python
df.loc[df['attendance'] < 0, 'attendance'] = np.nan
print(df['attendance'].describe())
```

```python
df.loc[df['assignments_completed'] < 0, 'assignments_completed'] = np.nan
print(df['assignments_completed'].describe())
```

```python
df.loc[df['previous_score'] < 0, 'previous_score'] = np.nan
print(df['previous_score'].describe())
```

```python
# To handle missing values by mean filling
df['study_hours'] = df['study_hours'].fillna(df['study_hours'].mean())
```

```python
df['attendance'] = df['attendance'].fillna(df['attendance'].mean())
```

```python
df['assignments_completed'] = df['assignments_completed'].fillna(df['assignments_completed'].mean())
```

```python
df['previous_score'] = df['previous_score'].fillna(df['previous_score'].mean())
```

```python
print(df.isnull().sum())
```

```python
# Creating New Two Features

# Feature 1 : Study hour ranges from 0 to 1, so covert in scale of 100.
df['study_time'] = df['study_hours'] * 10
df['study_time']
```

```python
# Feature 2: Calculating unser perform students
df['under_perform_students'] = np.where((df['study_hours'] < 4) & (df['attendance'] < 70) & (df['assignments_completed'] < 2), 1, 0)
df['under_perform_students'].value_counts()
```
